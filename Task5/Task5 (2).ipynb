```python
import pandas as pd
titanic = pd.read_csv("Titanic-Dataset.csv")
titanic
```

```python
titanic.isna().sum()
```

```python
avg_age = titanic["Age"].mean()
titanic["Age"] = titanic["Age"].fillna(avg_age)
titanic
```

```python
most_common_port = titanic["Embarked"].mode().iloc[0]
titanic["Embarked"] = titanic["Embarked"].fillna(most_common_port)
titanic
```

```python
titanic = titanic.drop(columns=["Cabin"])
titanic
```

```python
titanic.head()
```

```python
titanic.dtypes
```

```python
titanic["Age"] = pd.to_numeric(titanic["Age"])
titanic
```

```python
titanic.duplicated().sum()
```

```python
titanic["Age"] = titanic["Age"].astype("int64")
titanic
```

```python
titanic["Age"].describe()
```

```python
titanic["Age"].agg(["max", "min"])
```

```python
adult_passengers = titanic.query("Age >= 18 and Fare > 30")
adult_passengers
```

```python
adult_passengers.head()
```

```python
titanic["FamilySize"] = titanic["SibSp"] + titanic["Parch"] + 1
titanic
```

```python
titanic["FarePerPerson"] = titanic["Fare"] / titanic["FamilySize"]
titanic
```

```python
summary = titanic.groupby(["Pclass", "Sex"])[["Fare", "Survived"]].mean()
summary
```
